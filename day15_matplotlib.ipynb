# Day 15 — Matplotlib & Data Visualization
## Production OEE Dashboard Charts

Visualizing OEE and defect data from 240 production records.  
Charts exported as PNG files for portfolio use.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sample_data.csv")
df["availability"] = df["actual_run"] / df["planned"]
df["performance"]  = df["units"] / (df["planned"] * 1.8)
df["quality"]      = df["good_units"] / df["units"]
df["oee"]          = df["availability"] * df["performance"] * df["quality"]

print(f"Dataset ready: {df.shape[0]} records")
```

## Chart 1 — Average OEE by Machine Center
Bar chart showing fleet OEE performance against world class threshold.

```python
machine_oee = df.groupby("machine")["oee"].mean()

plt.figure(figsize=(10, 5))
plt.bar(machine_oee.index, machine_oee.values,
        color="steelblue", edgecolor="black")
plt.axhline(y=0.85, color="red", linestyle="--",
            linewidth=1.5, label="World Class (85%)")
plt.title("Average OEE by Machine Center", fontsize=14, fontweight="bold")
plt.xlabel("Machine")
plt.ylabel("Average OEE")
plt.ylim(0, 1.0)
plt.xticks(rotation=15, ha="right")
plt.legend()
plt.tight_layout()
plt.savefig("oee_by_machine.png", dpi=150)
plt.show()
print("Chart saved: oee_by_machine.png")
```

## Chart 2 — OEE Trend Over Time
Daily OEE trend for CNC Mill #4 with world class and minimum acceptable reference lines.

```python
cnc = df[df["machine"] == "CNC Mill #4"].copy()
cnc_daily = cnc.groupby("date")["oee"].mean().reset_index()

plt.figure(figsize=(12, 5))
plt.plot(cnc_daily["date"], cnc_daily["oee"],
         color="steelblue", linewidth=2, marker="o", markersize=4)
plt.axhline(y=0.85, color="red", linestyle="--",
            linewidth=1.5, label="World Class")
plt.axhline(y=0.70, color="orange", linestyle="--",
            linewidth=1.5, label="Minimum Acceptable")
plt.title("CNC Mill #4 — Daily OEE Trend", fontsize=14, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("OEE")
plt.ylim(0, 1.0)
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.savefig("oee_trend_cnc.png", dpi=150)
plt.show()
print("Chart saved: oee_trend_cnc.png")
```

## Chart 3 — Defect Code Pareto
Frequency of defect codes across all production records — NONE excluded.

```python
defect_counts = df[df["defect_code"] != "NONE"]["defect_code"].value_counts()

plt.figure(figsize=(10, 5))
bars = plt.barh(defect_counts.index, defect_counts.values,
                color="tomato", edgecolor="black")
for bar, val in zip(bars, defect_counts.values):
    plt.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2,
             str(val), va="center", fontsize=10)
plt.title("Defect Code Frequency — Pareto View",
          fontsize=14, fontweight="bold")
plt.xlabel("Occurrence Count")
plt.ylabel("Defect Code")
plt.tight_layout()
plt.savefig("defect_pareto.png", dpi=150)
plt.show()
print("Chart saved: defect_pareto.png")
```

## Key Findings

- OEE varies significantly across machine centers
- CNC Mill #4 trend shows day-over-day variability
- Defect codes distributed across DEF-001, DEF-004, DEF-007, DEF-011
